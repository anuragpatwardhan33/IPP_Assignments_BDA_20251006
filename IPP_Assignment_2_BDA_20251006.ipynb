```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

print("Libraries imported successfully.")
```

## Program 1 : Read and Display an Image Using OpenCV

```python

image_path = "flower.jpg"

if not os.path.exists(image_path):
    print(f"Error: {image_path} was not found. Place it beside this notebook.")
else:
    image = cv2.imread(image_path, cv2.IMREAD_COLOR)
    if image is None:
        print(f"Error: OpenCV could not read {image_path}.")
    else:
        print("Image dimensions (height, width, channels):", image.shape)
        print("Data type:", image.dtype)
        image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        plt.figure(figsize=(7, 5))
        plt.imshow(image_rgb)
        plt.title("Original Image")
        plt.axis("off")
        plt.show()
```

## Program 2 : Display Image Dimensions and Data Type

## Program 3 : Convert a Color Image to Grayscale

```python

image_path = "scene.jpg"

if not os.path.exists(image_path):
    print(f"Error: {image_path} was not found. Place it beside this notebook.")
else:
    image = cv2.imread(image_path, cv2.IMREAD_COLOR)
    if image is None:
        print(f"Error: OpenCV could not read {image_path}.")
    else:
        gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
        if cv2.imwrite("output_gray.jpg", gray_image):
            print("Grayscale image saved as output_gray.jpg")
            plt.figure(figsize=(6, 4))
            plt.imshow(gray_image, cmap="gray")
            plt.title("Grayscale Image")
            plt.axis("off")
            plt.show()
        else:
            print("Error: Could not save output_gray.jpg.")
```

## Program 4 : Display an Image Using Pillow

```python

image_path = "sunset.png"

if not os.path.exists(image_path):
    print(f"Error: {image_path} was not found. Place it beside this notebook.")
else:
    try:
        with Image.open(image_path) as image:
            print("Image size (width, height):", image.size)
            print("Image mode:", image.mode)
            plt.figure(figsize=(7, 5))
            plt.imshow(image)
            plt.title("Sunset Image")
            plt.axis("off")
            plt.show()
    except Exception as error:
        print("Error: Could not open image:", error)
```

## Program 5 : Resize an Image to 50%

```python

image_path = "scene.jpg"

if not os.path.exists(image_path):
    print(f"Error: {image_path} was not found. Place it beside this notebook.")
else:
    image = cv2.imread(image_path, cv2.IMREAD_COLOR)
    if image is None:
        print(f"Error: OpenCV could not read {image_path}.")
    else:
        height, width = image.shape[:2]
        new_width = max(1, int(width * 0.5))
        new_height = max(1, int(height * 0.5))
        resized_image = cv2.resize(
            image, (new_width, new_height), interpolation=cv2.INTER_AREA
        )
        print("Original dimensions (width x height):", width, "x", height)
        print("Resized dimensions (width x height):", new_width, "x", new_height)

        plt.figure(figsize=(10, 4))
        plt.subplot(1, 2, 1)
        plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
        plt.title("Original Image")
        plt.axis("off")
        plt.subplot(1, 2, 2)
        plt.imshow(cv2.cvtColor(resized_image, cv2.COLOR_BGR2RGB))
        plt.title("Resized Image (50%)")
        plt.axis("off")
        plt.tight_layout()
        plt.show()
```

## Program 6 : Crop the Center 200 × 200 Pixels

```python

image_path = "scene.jpg"

if not os.path.exists(image_path):
    print(f"Error: {image_path} was not found. Place it beside this notebook.")
else:
    image = cv2.imread(image_path, cv2.IMREAD_COLOR)
    if image is None:
        print(f"Error: OpenCV could not read {image_path}.")
    else:
        height, width = image.shape[:2]
        crop_size = 200
        if height < crop_size or width < crop_size:
            print("Error: Image must be at least 200 x 200 pixels.")
        else:
            x_start = (width - crop_size) // 2
            y_start = (height - crop_size) // 2
            cropped_image = image[
                y_start:y_start + crop_size,
                x_start:x_start + crop_size
            ]
            print("Cropped dimensions (height, width, channels):", cropped_image.shape)
            plt.figure(figsize=(5, 5))
            plt.imshow(cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB))
            plt.title("Center Crop (200 x 200)")
            plt.axis("off")
            plt.show()
```

## Program 7 : Rotate an Image 90° Clockwise

```python

image_path = "scene.jpg"

if not os.path.exists(image_path):
    print(f"Error: {image_path} was not found. Place it beside this notebook.")
else:
    image = cv2.imread(image_path, cv2.IMREAD_COLOR)
    if image is None:
        print(f"Error: OpenCV could not read {image_path}.")
    else:
        rotated_image = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)
        print("Original dimensions (width x height):", image.shape[1], "x", image.shape[0])
        print("Rotated dimensions (width x height):", rotated_image.shape[1], "x", rotated_image.shape[0])

        plt.figure(figsize=(10, 4))
        plt.subplot(1, 2, 1)
        plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
        plt.title("Original Image")
        plt.axis("off")
        plt.subplot(1, 2, 2)
        plt.imshow(cv2.cvtColor(rotated_image, cv2.COLOR_BGR2RGB))
        plt.title("Rotated 90° Clockwise")
        plt.axis("off")
        plt.tight_layout()
        plt.show()
```

## Program 8 : Flip an Image Horizontally

```python

image_path = "car.jpg"

if not os.path.exists(image_path):
    print(f"Error: {image_path} was not found. Place it beside this notebook.")
else:
    image = cv2.imread(image_path, cv2.IMREAD_COLOR)
    if image is None:
        print(f"Error: OpenCV could not read {image_path}.")
    else:
        flipped_image = cv2.flip(image, 1)
        print("Image flipped horizontally.")
        plt.figure(figsize=(10, 4))
        plt.subplot(1, 2, 1)
        plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
        plt.title("Original Image")
        plt.axis("off")
        plt.subplot(1, 2, 2)
        plt.imshow(cv2.cvtColor(flipped_image, cv2.COLOR_BGR2RGB))
        plt.title("Horizontally Flipped Image")
        plt.axis("off")
        plt.tight_layout()
        plt.show()
```

## Program 9 : Convert an Image to HSV and Display H, S, V

```python

image_path = "scene.jpg"

if not os.path.exists(image_path):
    print(f"Error: {image_path} was not found. Place it beside this notebook.")
else:
    image = cv2.imread(image_path, cv2.IMREAD_COLOR)
    if image is None:
        print(f"Error: OpenCV could not read {image_path}.")
    else:
        hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
        hue, saturation, value = cv2.split(hsv_image)
        print("Converted image to HSV and extracted H, S, and V channels.")

        plt.figure(figsize=(12, 7))
        plt.subplot(2, 2, 1)
        plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
        plt.title("Original Image")
        plt.axis("off")
        plt.subplot(2, 2, 2)
        plt.imshow(hue, cmap="gray")
        plt.title("Hue (H)")
        plt.axis("off")
        plt.subplot(2, 2, 3)
        plt.imshow(saturation, cmap="gray")
        plt.title("Saturation (S)")
        plt.axis("off")
        plt.subplot(2, 2, 4)
        plt.imshow(value, cmap="gray")
        plt.title("Value (V)")
        plt.axis("off")
        plt.tight_layout()
        plt.show()
```

## Program 10 : Calculate Mean Squared Error (MSE)

```python

image1_path = "img1.jpg"
image2_path = "img2.jpg"

if not os.path.exists(image1_path):
    print(f"Error: {image1_path} was not found. Place it beside this notebook.")
elif not os.path.exists(image2_path):
    print(f"Error: {image2_path} was not found. Place it beside this notebook.")
else:
    image1 = cv2.imread(image1_path, cv2.IMREAD_GRAYSCALE)
    image2 = cv2.imread(image2_path, cv2.IMREAD_GRAYSCALE)

    if image1 is None or image2 is None:
        print("Error: One or both images could not be read.")
    elif image1.shape != image2.shape:
        print("Error: Both images must have the same dimensions.")
        print("Image 1 dimensions:", image1.shape)
        print("Image 2 dimensions:", image2.shape)
    else:
        difference = image1.astype(np.float64) - image2.astype(np.float64)
        mse = np.mean(difference ** 2)
        print("Image dimensions (height, width):", image1.shape)
        print(f"Mean Squared Error (MSE): {mse:.4f}")
```
